Find the Largest Number

```python
def largest_n(num):
    max_num=num[0]
    for i in num:
        if i> max_num:
           max_num=i
    print(max_num)       
   

numbers = [10, 45, 23, 89, 12]
largest_n(numbers)
```

Count Vowels

```python
def vowels(w):
    d='aeiouAEIOU'
    count=0
    for i in w:
        if i in d:
            count+=1
    return count

    
text = "python programming"
vowels(text)
```

Find Prime Numbers

```python
def prime_n(n):
    for num in n:
        if num <2:
            continue

        prime =True

        for i in range(2,num):
            if num % i ==0:
                prime=False
                break
        if prime:
             print(num)

  
    
numbers = [2, 4, 7, 9, 11, 15, 17, 20]
prime_n(numbers)
```

```python
def remove(n):
    unique_n=[]
    for i in n:
        if i not in unique_n:
            unique_n=unique_n+[i]

    print (unique_n)
numbers = [1, 2, 2, 3, 4, 4, 5, 1]
remove(numbers)

```

Character Frequency

```python
def fre_c(t):
    txt={}

    for i in t:
        if i in txt:
            txt[i]+=1
        else:
            txt[i]=1
    for i in txt:
        print(i,":",txt[i])
                     

#text= "programming"
text=input()

fre_c(text)
```
